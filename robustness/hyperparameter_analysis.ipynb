```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast

plt.style.use('default')
```

```python
sns.set_palette("husl")
```

```python
df = pd.read_csv('data/methods.tsv', sep='\t', header=None)
df.columns = ['Unnamed: 0', 'precision', 'recall', 'f1-score', 'support', 'split', 'method', 'hash', 'method_params', 'task', 'epoch']
df
```

```python
df = df[df['Unnamed: 0'] == 'accuracy'].copy()
df
```

```python
df['accuracy'] = pd.to_numeric(df['precision'])
df
```

```python
def parse_params(params_str):
    try:
        params_str = params_str.strip('"')
        params = ast.literal_eval(params_str)
        return params
    except:
        return {}
```

```python
df['parsed_params'] = df['method_params'].apply(parse_params)
df['method_params'] = df['method_params'].apply(lambda x: str(x) if not isinstance(x, str) else x)
df['lr'] = df['parsed_params'].apply(lambda x: x.get('lr', np.nan))
df['batch_size'] = df['parsed_params'].apply(lambda x: x.get('batch_size', np.nan))
df['scoring'] = df['parsed_params'].apply(lambda x: x.get('scoring', np.nan))
df
```

```python
# Only keep configs with all 5 splits
split_counts = df.groupby(['hash', 'method', 'task', 'epoch', 'method_params'])['split'].nunique().reset_index()
full_configs = split_counts[split_counts['split'] == 5]
df_full = pd.merge(df, full_configs.drop('split', axis=1), on=['hash', 'method', 'task', 'epoch', 'method_params'])
df_full
```

```python
# Compute mean and std of accuracy per config
agg = df_full.groupby(['hash', 'method', 'task', 'epoch', 'method_params'])['accuracy'].agg(['mean', 'std']).reset_index()
agg = agg.rename(columns={'mean': 'accuracy_mean', 'std': 'accuracy_std'})
df_full = pd.merge(df_full, agg[['hash', 'method', 'task', 'epoch', 'method_params', 'accuracy_std']], on=['hash', 'method', 'task', 'epoch', 'method_params'], how='left')
df_full
```

```python
def plot_hyper_param_analysis_per_task(df_full, tasks, name):
    df_filtered = df_full[df_full['task'].isin(tasks)]

    # learning rate
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        lr_means = method_data.groupby('lr')['accuracy'].mean()
        lr_stds = method_data.groupby('lr')['accuracy'].std()
        ax.errorbar(lr_means.index, lr_means.values, yerr=lr_stds.values, marker='o', label=label, alpha=0.8) #, linewidth=0.5, markersize=4)
        ax.set_xlabel('Learning Rate')
        ax.set_ylabel('Mean Accuracy')
        ax.set_title('Learning Rate Impact')
        ax.legend()
        ax.set_xscale('log')
        all_lrs = sorted(df_filtered['lr'].unique())
        ax.set_xticks(all_lrs)
        ax.grid(True, alpha=0.3, zorder=0)

    plt.savefig(f'figures/{name}_lr_impact.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # batch size
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        batch_means = method_data.groupby('batch_size')['accuracy'].mean()
        batch_stds = method_data.groupby('batch_size')['accuracy'].std()
        ax.errorbar(batch_means.index, batch_means.values, yerr=batch_stds.values, marker='o', label=label, alpha=0.8) #, linewidth=0.5, markersize=4)
        ax.set_xlabel('Batch size')
        ax.set_ylabel('Mean Accuracy')
        ax.set_title('Batch Size Impact')
        ax.legend()
        all_bs = sorted(df_filtered['batch_size'].unique())
        ax.set_xticks(all_bs)
        ax.grid(True, alpha=0.3, zorder=0)

    plt.savefig(f'figures/{name}_batch_size_impact.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)


    # pooling functions
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    scoring_methods = ['gated_attn', 'attn', 'mean', 'max']
    x = np.arange(len(scoring_methods))
    width = 0.35
    for i, method in enumerate(['multigrate_mil', 'multigrate_mil_reg']):
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        scoring_means = method_data.groupby('scoring')['accuracy'].mean().reindex(scoring_methods)
        scoring_stds = method_data.groupby('scoring')['accuracy'].std().reindex(scoring_methods)
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        bars = ax.bar(x + i*width, scoring_means.values, width, label=label, alpha=0.8, zorder=3)
        for j, (bar, std) in enumerate(zip(bars, scoring_stds.values)):
            ax.errorbar(bar.get_x() + bar.get_width()/2, bar.get_height(), yerr=std, fmt='none', color='black', linewidth=0.8, zorder=4) # capsize=3, 
        ax.set_xlabel('Pooling Function')
        ax.set_ylabel('Mean Accuracy')
        ax.set_title('Pooling Function Impact')
        ax.set_xticks(x + width/2)
        ax.set_xticklabels(scoring_methods, rotation=45)
        ax.legend()
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_pooling_function_impact.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # epoch
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        epoch_means = method_data.groupby('epoch')['accuracy'].mean()
        epoch_means.index = epoch_means.index.astype(float).astype(int)
        epoch_means = epoch_means.sort_index()
        epoch_stds = method_data.groupby('epoch')['accuracy'].std()
        epoch_stds.index = epoch_stds.index.astype(float).astype(int)
        epoch_stds = epoch_stds.loc[epoch_means.index]
        ax.errorbar(epoch_means.index, epoch_means.values, yerr=epoch_stds.values, marker='o', label=label, alpha=0.8) #, linewidth=0.5, markersize=4)
        ax.set_xlabel('Epoch')
        ax.set_ylabel('Mean Accuracy')
        ax.set_title('Epoch Impact')
        ax.legend()
        ax.set_xticks(epoch_means.index)
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_epoch_impact.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # lr stability
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        # Use accuracy_std for stability
        lr_stability = method_data.groupby('lr')['accuracy_std'].mean()
        ax.plot(lr_stability.index, lr_stability.values, marker='o', label=label, alpha=0.8)
        ax.set_xlabel('Learning Rate')
        ax.set_ylabel('Mean Std')
        ax.set_title('Learning Rate vs Stability')
        ax.legend()
        ax.set_xscale('log')
        all_lrs = sorted(df_filtered['lr'].unique())
        ax.set_xticks(all_lrs)
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_lr_stability.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # batch size stability
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        # Use accuracy_std for stability
        batch_stability = method_data.groupby('batch_size')['accuracy_std'].mean()
        ax.plot(batch_stability.index, batch_stability.values, marker='o', label=label, alpha=0.8)
        ax.set_xlabel('Batch Size')
        ax.set_ylabel('Mean Std')
        ax.set_title('Batch Size vs Stability')
        ax.legend()
        all_bs = sorted(df_filtered['batch_size'].unique())
        ax.set_xticks(all_bs)
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_batch_size_stability.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # pooling function stability
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for i, method in enumerate(['multigrate_mil', 'multigrate_mil_reg']):
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        scoring_stability = method_data.groupby('scoring')['accuracy_std'].mean().reindex(scoring_methods)
        bars = ax.bar(x + i*width, scoring_stability.values, width, label=label, alpha=0.8, zorder=3)
        ax.set_xlabel('Scoring Method')
        ax.set_ylabel('Mean Std')
        ax.set_title('Scoring Method vs Stability')
        ax.set_xticks(x + width/2)
        ax.set_xticklabels(scoring_methods, rotation=45)
        ax.legend()
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_pooling_function_stability.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)

    # epoch stability
    fig, ax = plt.subplots(1, 1, figsize=(2, 2))
    for method in ['multigrate_mil', 'multigrate_mil_reg']:
        label = 'MultiMIL classification' if method == 'multigrate_mil' else 'MultiMIL regression'
        method_data = df_filtered[df_filtered['method'] == method]
        epoch_stability = method_data.groupby('epoch')['accuracy_std'].mean()
        epoch_stability.index = epoch_stability.index.astype(float).astype(int)
        epoch_stability = epoch_stability.sort_index()
        ax.plot(epoch_stability.index, epoch_stability.values, marker='o', label=label, alpha=0.8)
        ax.set_xlabel('Epoch')
        ax.set_ylabel('Mean Std')
        ax.set_title('Epoch vs Stability')
        ax.legend()
        ax.set_xticks(epoch_stability.index)
        ax.grid(True, alpha=0.3, zorder=0)
    plt.savefig(f'figures/{name}_epoch_stability.svg', bbox_inches='tight', dpi=300)
    plt.close(fig)
```

```python
tasks = ['pbmc_scvi', 'pbmc_scgpt', 'pbmc_pca', 'pbmc_geneformer']
for task in tasks:
    plot_hyper_param_analysis_per_task(df_full, tasks=[task], name=task)
    
```
