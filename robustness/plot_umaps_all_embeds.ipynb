```python
import scanpy as sc
import pandas as pd
import seaborn as sns
sc.set_figure_params(vector_friendly=True, figsize=(4,4), dpi=300)
```

```python
adata = sc.read('/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/multigrate_mil_reg/pbmc_scvi/6e49b5e194_adata_both.h5ad')
adata
```

```python
celltype_col = 'initial_clustering'
celltype_counts = adata.obs[celltype_col].value_counts()
celltype_order = list(celltype_counts.index)
celltype_colors = sns.color_palette('tab20', len(celltype_order))
celltype_palette = dict(zip(celltype_order, celltype_colors))
celltype_palette
```

```python
status_palette = {
    'Healthy': '#C2BC00',
    'Mild': '#E59A88',
    'Severe': '#B62A6F'
}
```

```python
path_dict = {
    'scVI': '/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/multigrate_mil_reg/pbmc_scvi/6e49b5e194_adata_both.h5ad',
    'scGPT': '/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/multigrate_mil_reg/pbmc_scgpt/9f082329d4_adata_both.h5ad',
    'GeneFormer': '/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/multigrate_mil_reg/pbmc_geneformer/5d1dd303a8_adata_both.h5ad',
    'PCA': '/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/multigrate_mil_reg/pbmc_pca/4288a543ba_adata_both.h5ad',
}
```

```python
for method, adata_path in path_dict.items():
    adata = sc.read_h5ad(adata_path)
    if 'cell_attn' in adata.obs.columns: # if best method did not use cell attention, skip
        sc.pl.umap(adata, color=['cell_attn_0', 'cell_attn_1', 'cell_attn_2', 'cell_attn_3', 'cell_attn_4', 'cell_attn'], vmax='p99', color_map='GnBu', frameon=False, ncols=1, save=f'_{method}_reg_cell_attn.svg', show=False)
    sc.pl.umap(adata, color=['initial_clustering'], palette=celltype_palette, frameon=False, ncols=1, save=f'_{method}_reg_initial_clustering.svg', show=False)
    sc.pl.umap(adata, color=['Status_on_day_collection_summary'], palette=status_palette, frameon=False, ncols=1, save=f'_{method}_reg_status.svg', show=False)
    sc.pl.umap(adata, color=['Site'], frameon=False, ncols=1, save=f'_{method}_reg_site.svg', show=False)
```
