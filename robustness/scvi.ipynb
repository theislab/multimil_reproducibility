```python
import scanpy as sc
import scvi
import seaborn as sns
import torch
```

```python
adata = sc.read('/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/pp/pbmc_full_rna.h5ad')
adata
```

```python
scvi.settings.seed = 0
print("Last run with scvi-tools version:", scvi.__version__)
```

```python
torch.set_float32_matmul_precision("high")
```

```python
scvi.model.SCVI.setup_anndata(adata, layer="counts", batch_key="sample_id", categorical_covariate_keys=["Site"])
```

```python
model = scvi.model.SCVI(adata, n_layers=2, n_latent=30, gene_likelihood="nb")
```

```python
model.train()
```

```python
SCVI_LATENT_KEY = "X_scVI"
adata.obsm[SCVI_LATENT_KEY] = model.get_latent_representation()
```

```python
embed = sc.AnnData(adata.obsm[SCVI_LATENT_KEY], obs=adata.obs)
embed
```

```python
sc.pp.neighbors(embed)
```

```python
sc.tl.umap(embed)
```

```python
sc.pl.umap(
    embed,
    color=embed.obs.columns,
    frameon=False,
    ncols=1,
)
```

```python
embed.write('data/scvi_latent.h5ad')
```
