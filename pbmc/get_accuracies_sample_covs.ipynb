```python
import pandas as pd
```

```python
df = pd.read_csv('/lustre/groups/ml01/projects/2022_multigrate_anastasia.litinetskaya/multimil_reproducibility/pipeline_mil/data/reports/methods.tsv', sep='\t')
df
```

```python
df = df[df['task'] == 'pbmc_scvi_sample_covs'].copy()
df
```

```python
df = df[df['Unnamed: 0'] == 'accuracy'].copy()
df
```

```python
df_ = df[['split', 'epoch', 'hash']].groupby(['epoch', 'hash']).count()
df_ = df_[df_['split'] == 5].copy()
df_
```

```python
df_.reset_index(inplace=True)
```

```python
df_['epoch_hash'] = df_['epoch'].astype(str) + '_' + df_['hash']
df_
```

```python
df['epoch_hash'] = df['epoch'].astype(str) + '_' + df['hash']
```

```python
df = df[df['epoch_hash'].isin(df_['epoch_hash'])]
df
```

```python
df = df.groupby(['hash', 'epoch']).agg({'precision': ['mean', 'std'], 'method_params': 'first', 'epoch_hash': 'first'}).reset_index().sort_values(by=('precision', 'mean'), ascending=False)
df
```

```python
df.to_csv('sample_covs_accuracies.tsv', sep='\t', index=False)
```
